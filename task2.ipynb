```python
import numpy as np
import k3d
from source import Source
from visualize import build_sphere

src = Source(Y=10000)
E = 15.0
dirs = src.emit(E, seed=0)          # (n, 3) единичные направления
print(f'E = {E} МэВ -> n = {len(dirs)} фотонов')

# Направления уже единичные: показываем первые 2000 концов на единичной сфере.
n_show = min(2000, len(dirs))

plot = k3d.plot()
plot += build_sphere(radius=1.0, opacity=0.2, color=0x3366aa)   # единичная сфера
plot += k3d.points(dirs[:n_show], color=0x22dd22, point_size=0.03, shader='3d')
plot.display()
```
